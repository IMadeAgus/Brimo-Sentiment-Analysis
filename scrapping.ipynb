```python
from google_play_scraper import reviews
import pandas as pd
import os
```

```python
# Mengambil ulasan
scrapreview, _ = reviews(
    'id.co.bri.brimo',
    lang='id',
    country='id',
    count=50000
)
```

```python
app_reviews_df = pd.DataFrame(scrapreview)
app_reviews_df.head()
```

```python
# os.mkdir('dataset')
folder_path = f'dataset'
os.makedirs(folder_path, exist_ok=True)
app_reviews_df.to_csv(f'{folder_path}/review_app_brimo.csv', index=False)
```
